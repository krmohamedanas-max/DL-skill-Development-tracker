```python
# Import PyTorch for GPU tensor computations.
import torch

# Import neural network utilities.
import torch.nn as nn

# Import time module for benchmarking execution speed.
import time
```

```python
# Select CUDA if a GPU is available.
# Otherwise, use the CPU.

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Running on:", device)
```

```python
# Implement the Swish activation function.
#
# Swish Equation:
# f(x) = x × sigmoid(beta × x)
#
# beta controls the shape of the activation curve.

class Swish(nn.Module):

    def __init__(self, beta=1.0):

        super().__init__()

        self.beta = beta

    def forward(self, x):

        # Apply the Swish activation.
        return x * torch.sigmoid(self.beta * x)
```

```python
# Create the custom Swish activation layer.

swish = Swish(beta=1.0).to(device)
```

```python
# Generate a large tensor to evaluate
# runtime performance on the GPU.

input_tensor = torch.randn(

    1000,

    1000,

    device=device

)
```

```python
# Apply the custom Swish activation
# to the input tensor.

output = swish(input_tensor)

print("Output Shape :", output.shape)
```

```python
# Measure execution time of
# the custom Swish implementation.

start = time.time()

for _ in range(100):

    output = swish(input_tensor)

# Synchronize CUDA before stopping the timer.
if device.type == "cuda":

    torch.cuda.synchronize()

end = time.time()

print("Custom Swish Time :", end - start, "seconds")
```

```python
# Compare runtime with PyTorch's
# built-in SiLU activation.

silu = nn.SiLU().to(device)

start = time.time()

for _ in range(100):

    reference = silu(input_tensor)

if device.type == "cuda":

    torch.cuda.synchronize()

end = time.time()

print("PyTorch SiLU Time :", end - start, "seconds")
```

```python
# Verify that both implementations
# produce nearly identical outputs.

difference = torch.mean(

    torch.abs(

        output - reference

    )

)

print("Average Difference :", difference.item())
```
